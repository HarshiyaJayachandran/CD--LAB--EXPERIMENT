```python
%%writefile symtab.l

%{
#include <stdio.h>
#include <string.h>
#include <stdlib.h>

struct symtab {
    char name[30];
    int type;
} symtab[100];

int sc = 0;

int lookup(char *s)
{
    int i;

    for (i = 0; i < sc; i++)
    {
        if (strcmp(symtab[i].name, s) == 0)
            return i;
    }

    return -1;
}

void insert(char *s)
{
    if (lookup(s) == -1)
    {
        strcpy(symtab[sc].name, s);
        symtab[sc].type = 1;
        sc++;
    }
}
%}

DIGIT [0-9]
ID [a-zA-Z_][a-zA-Z0-9_]*

%%

"/*"([^*]|\*+[^*/])*\*+"/" {
    printf("Comment : %s\n", yytext);
}

"//".* {
    printf("Comment : %s\n", yytext);
}

{ID} {
    insert(yytext);
    printf("Identifier : %s\n", yytext);
}

{DIGIT}+ {
    printf("Constant : %s\n", yytext);
}

"+"|"-"|"*"|"/"|"="|"<"|">" {
    printf("Operator : %s\n", yytext);
}

[ \t\n] {
    /* skip whitespace */
}

. {
    /* ignore other characters */
}

%%

int yywrap()
{
    return 1;
}

int main(int argc, char *argv[])
{
    if (argc < 2)
    {
        printf("Usage: %s <input file>\n", argv[0]);
        return 1;
    }

    yyin = fopen(argv[1], "r");

    if (!yyin)
    {
        printf("Cannot open file %s\n", argv[1]);
        return 1;
    }

    yylex();

    printf("\nSYMBOL TABLE\n");
    printf("S.No\tName\n");

    int i;

    for (i = 0; i < sc; i++)
    {
        printf("%d\t%s\n", i + 1, symtab[i].name);
    }

    fclose(yyin);

    return 0;
}
```

```python
%%writefile lexer.l

%{
#include <stdio.h>
%}

KEYWORD int|float|char|double|void|for|while|if|else|return|struct|switch|case|break|do

%%

"#include" {
    printf("Preprocessor Directive : %s\n", yytext);
}

"<"[a-zA-Z.]+">" {
    printf("Header File : %s\n", yytext);
}

{KEYWORD} {
    printf("Keyword : %s\n", yytext);
}

[a-zA-Z_][a-zA-Z0-9_]* {
    printf("Identifier : %s\n", yytext);
}

[0-9]+ {
    printf("Number : %s\n", yytext);
}

"=="|"<="|">="|"!=" {
    printf("Operator : %s\n", yytext);
}

"+"|"-"|"*"|"/"|"="|"<"|">" {
    printf("Operator : %s\n", yytext);
}

[(){};,] {
    printf("Delimiter : %s\n", yytext);
}

[ \t\n] {
    /* Skip whitespace */
}

. {
    /* Ignore other characters */
}

%%

int yywrap()
{
    return 1;
}

int main(int argc, char *argv[])
{
    if (argc < 2)
    {
        printf("Usage: %s <input file>\n", argv[0]);
        return 1;
    }

    yyin = fopen(argv[1], "r");

    if (!yyin)
    {
        printf("Cannot open file %s\n", argv[1]);
        return 1;
    }

    yylex();

    printf("\nEnd of file\n");

    fclose(yyin);

    return 0;
}
```

```python
%%writefile art_expr.l

%{
#include <stdio.h>
#include "art_expr.tab.h"
%}

%%

[a-zA-Z][0-9a-zA-Z]*    { return ID; }

[0-9]+                  { return DIG; }

[ \t]+                  { /* ignore spaces */ }

\n                      { return 0; }

.                       { return yytext[0]; }

%%

int yywrap()
{
    return 1;
}
```

```python
%%writefile art_expr.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex(void);
int yyerror(const char *s);
%}

%token ID DIG

%left '+' '-'
%left '*' '/'

%right UMINUS

%%

stmt:
    expn
    ;

expn:
      expn '+' expn
    | expn '-' expn
    | expn '*' expn
    | expn '/' expn
    | '-' expn %prec UMINUS
    | '(' expn ')'
    | DIG
    | ID
    ;

%%

int main()
{
    printf("Enter the Expression:\n");

    if (yyparse() == 0)
    {
        printf("Valid Expression\n");
    }

    return 0;
}

int yyerror(const char *s)
{
    printf("Invalid Expression\n");
    return 0;
}
```

```python
%%writefile valvar.l

%{
#include "valvar.tab.h"
%}

%%

[a-zA-Z] {
    return LET;
}

[0-9] {
    return DIG;
}

[ \t] {
    /* Ignore spaces and tabs */
}

\n {
    return 0;
}

. {
    return yytext[0];
}

%%

int yywrap()
{
    return 1;
}
```

```python
%%writefile valvar.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex(void);
int yyerror(const char *s);
%}

%token LET DIG

%%

variable:
    var
    ;

var:
      var DIG
    | var LET
    | LET
    ;

%%

int main()
{
    printf("Enter the variable:\n");

    if (yyparse() == 0)
    {
        printf("Valid variable\n");
    }

    return 0;
}

int yyerror(const char *s)
{
    printf("Invalid variable\n");
    return 0;
}
```
