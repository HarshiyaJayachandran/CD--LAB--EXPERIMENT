```python
%%writefile control.l

%{
#include "control.tab.h"
%}

%%

"if"        { return IF; }
"else"      { return ELSE; }
"for"       { return FOR; }
"while"     { return WHILE; }
"switch"    { return SWITCH; }
"case"      { return CASE; }
"default"   { return DEFAULT; }

[a-zA-Z_][a-zA-Z0-9_]*   { return ID; }

[0-9]+      { return NUM; }

"{"         { return LBRACE; }
"}"         { return RBRACE; }
"("         { return LPAREN; }
")"         { return RPAREN; }
":"         { return COLON; }
";"         { return SEMICOLON; }

"=="        { return EQ; }
"<="        { return LE; }
">="        { return GE; }
"<"         { return LT; }
">"         { return GT; }
"="         { return ASSIGN; }

[ \t\n]+    { /* skip whitespace */ }

.           { return yytext[0]; }

%%

int yywrap()
{
    return 1;
}
```

```python
%%writefile control.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex(void);
int yyerror(const char *s);
%}

%token IF ELSE FOR WHILE SWITCH CASE DEFAULT
%token ID NUM
%token LBRACE RBRACE LPAREN RPAREN COLON SEMICOLON
%token EQ LE GE LT GT ASSIGN

%%

program:
    stmt_list
    ;

stmt_list:
      stmt_list stmt
    | stmt
    ;

stmt:
      if_stmt
    | while_stmt
    | for_stmt
    | switch_stmt
    ;

if_stmt:
      IF LPAREN cond RPAREN block
    | IF LPAREN cond RPAREN block ELSE block
    ;

while_stmt:
    WHILE LPAREN cond RPAREN block
    ;

for_stmt:
    FOR LPAREN ID ASSIGN NUM SEMICOLON cond SEMICOLON ID ASSIGN ID RPAREN block
    ;

switch_stmt:
    SWITCH LPAREN ID RPAREN LBRACE case_list RBRACE
    ;

case_list:
      case_list CASE NUM COLON simple_stmt
    | case_list DEFAULT COLON simple_stmt
    | CASE NUM COLON simple_stmt
    | DEFAULT COLON simple_stmt
    ;

block:
    LBRACE simple_stmt_list RBRACE
    ;

simple_stmt_list:
      simple_stmt_list simple_stmt
    | simple_stmt
    ;

simple_stmt:
    ID ASSIGN NUM SEMICOLON
    ;

cond:
    ID relop NUM
    ;

relop:
      EQ
    | LE
    | GE
    | LT
    | GT
    ;

%%

int main()
{
    printf("Enter a C control structure syntax:\n");

    if (yyparse() == 0)
    {
        printf("Valid control structure syntax.\n");
    }

    return 0;
}

int yyerror(const char *s)
{
    printf("Invalid control structure syntax.\n");
    return 0;
}
```

```python
%%writefile cal.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex(void);
int yyerror(const char *s);
%}

%define api.value.type {double}

%token NUM

%left '+' '-'
%left '*' '/'
%right UMINUS

%%

input:
      /* empty */
    | input line
    ;

line:
      '\n'
    | E '\n' {
          printf("Answer: %g\n", $1);
      }
    ;

E:
      E '+' E {
          $$ = $1 + $3;
      }
    | E '-' E {
          $$ = $1 - $3;
      }
    | E '*' E {
          $$ = $1 * $3;
      }
    | E '/' E {
          if ($3 == 0)
          {
              printf("Error: Division by zero\n");
              $$ = 0;
          }
          else
          {
              $$ = $1 / $3;
          }
      }
    | '-' E %prec UMINUS {
          $$ = -$2;
      }
    | '(' E ')' {
          $$ = $2;
      }
    | NUM {
          $$ = $1;
      }
    ;

%%

int main()
{
    printf("Enter arithmetic expressions:\n");

    yyparse();

    return 0;
}

int yyerror(const char *s)
{
    printf("Invalid expression\n");
    return 0;
}
```

```python
%%writefile tac.y

%{
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

int yylex(void);
int yyerror(const char *s);

int tempCount = 1;
%}

%union {
    char *str;
}

%token <str> ID NUM
%type <str> expr

%left '+' '-'
%left '*' '/'

%%

stmt:
    ID '=' expr
    {
        printf("%s = %s\n", $1, $3);
        free($1);
        free($3);
    }
    ;

expr:
      expr '+' expr
      {
          char temp[20];

          sprintf(temp, "t%d", tempCount++);

          printf("%s = %s + %s\n", temp, $1, $3);

          $$ = strdup(temp);

          free($1);
          free($3);
      }

    | expr '-' expr
      {
          char temp[20];

          sprintf(temp, "t%d", tempCount++);

          printf("%s = %s - %s\n", temp, $1, $3);

          $$ = strdup(temp);

          free($1);
          free($3);
      }

    | expr '*' expr
      {
          char temp[20];

          sprintf(temp, "t%d", tempCount++);

          printf("%s = %s * %s\n", temp, $1, $3);

          $$ = strdup(temp);

          free($1);
          free($3);
      }

    | expr '/' expr
      {
          char temp[20];

          sprintf(temp, "t%d", tempCount++);

          printf("%s = %s / %s\n", temp, $1, $3);

          $$ = strdup(temp);

          free($1);
          free($3);
      }

    | ID
      {
          $$ = $1;
      }

    | NUM
      {
          $$ = $1;
      }
    ;

%%

int main()
{
    printf("Enter the expression:\n");

    yyparse();

    return 0;
}

int yyerror(const char *s)
{
    printf("Error: %s\n", s);

    return 0;
}
```

```python
%%writefile typecheck.y

%{
#include <stdio.h>
#include <string.h>
#include <stdlib.h>

int yylex(void);
int yyerror(const char *s);

struct sym
{
    char name[20];
    char type[10];
};

struct sym table[50];
int n = 0;

void insert(char *name, char *type)
{
    int i;

    /* Check whether variable already exists */
    for (i = 0; i < n; i++)
    {
        if (strcmp(table[i].name, name) == 0)
        {
            strcpy(table[i].type, type);
            return;
        }
    }

    if (n < 50)
    {
        strcpy(table[n].name, name);
        strcpy(table[n].type, type);
        n++;
    }
}

char *typeOf(char *name)
{
    int i;

    for (i = 0; i < n; i++)
    {
        if (strcmp(table[i].name, name) == 0)
            return table[i].type;
    }

    return "undefined";
}
%}

%union
{
    char *str;
}

%token <str> ID NUM
%token INT FLOAT

%type <str> expr

%left '+' '-'
%left '*' '/'

%%

program:
    stmts
    ;

stmts:
      stmts stmt
    | stmt
    ;

stmt:
      decl
    | assign
    ;

decl:
      INT ID ';'
      {
          insert($2, "int");
          printf("Declared %s as int\n", $2);
          free($2);
      }

    | FLOAT ID ';'
      {
          insert($2, "float");
          printf("Declared %s as float\n", $2);
          free($2);
      }
    ;

assign:
    ID '=' expr ';'
    {
        char *lt = typeOf($1);

        if (strcmp(lt, "undefined") == 0)
        {
            printf("Undefined variable: %s\n", $1);
        }
        else if (strcmp($3, "mismatch") == 0)
        {
            printf("Type mismatch in assignment to %s\n", $1);
        }
        else if (strcmp(lt, $3) == 0)
        {
            printf("No type mismatch in expression: %s = ...\n", $1);
        }
        else
        {
            printf("Type mismatch in assignment to %s\n", $1);
        }

        free($1);
        free($3);
    }
    ;

expr:
      ID
      {
          char *t = typeOf($1);

          if (strcmp(t, "undefined") == 0)
              printf("Undefined variable: %s\n", $1);

          $$ = strdup(t);
          free($1);
      }

    | NUM
      {
          $$ = strdup("int");
          free($1);
      }

    | expr '+' expr
      {
          if (strcmp($1, $3) == 0)
              $$ = strdup($1);
          else
              $$ = strdup("mismatch");

          free($1);
          free($3);
      }

    | expr '-' expr
      {
          if (strcmp($1, $3) == 0)
              $$ = strdup($1);
          else
              $$ = strdup("mismatch");

          free($1);
          free($3);
      }

    | expr '*' expr
      {
          if (strcmp($1, $3) == 0)
              $$ = strdup($1);
          else
              $$ = strdup("mismatch");

          free($1);
          free($3);
      }

    | expr '/' expr
      {
          if (strcmp($1, $3) == 0)
              $$ = strdup($1);
          else
              $$ = strdup("mismatch");

          free($1);
          free($3);
      }
    ;

%%

int main()
{
    printf("Enter declarations and expressions:\n");

    yyparse();

    return 0;
}

int yyerror(const char *s)
{
    printf("Syntax Error: %s\n", s);
    return 0;
}
```

```python
%%writefile optimize.y

%{
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <ctype.h>

int yylex(void);
int yyerror(const char *s);

int is_number(const char *s)
{
    if (s == NULL || *s == '\0')
        return 0;

    for (int i = 0; s[i] != '\0'; i++)
    {
        if (!isdigit((unsigned char)s[i]))
            return 0;
    }

    return 1;
}
%}

%union {
    char *str;
}

%token <str> ID NUM
%type <str> expr

%left '+' '-'
%left '*' '/'

%%

stmt_list:
      stmt_list stmt
    | stmt
    ;

stmt:
    ID '=' expr ';'
    {
        printf("%s = %s\n", $1, $3);

        free($1);
        free($3);
    }
    ;

expr:
      NUM
      {
          $$ = $1;
      }

    | ID
      {
          $$ = $1;
      }

    | expr '+' expr
      {
          char buf[100];

          if (is_number($1) && is_number($3))
          {
              int result = atoi($1) + atoi($3);

              sprintf(buf, "%d", result);

              printf("// Constant Folding: %s + %s -> %s\n",
                     $1, $3, buf);

              $$ = strdup(buf);
          }
          else if (strcmp($3, "0") == 0)
          {
              printf("// Algebraic Simplification: x + 0 -> x\n");

              $$ = strdup($1);
          }
          else if (strcmp($1, "0") == 0)
          {
              printf("// Algebraic Simplification: 0 + x -> x\n");

              $$ = strdup($3);
          }
          else
          {
              sprintf(buf, "%s + %s", $1, $3);
              $$ = strdup(buf);
          }

          free($1);
          free($3);
      }

    | expr '-' expr
      {
          char buf[100];

          if (is_number($1) && is_number($3))
          {
              int result = atoi($1) - atoi($3);

              sprintf(buf, "%d", result);

              printf("// Constant Folding: %s - %s -> %s\n",
                     $1, $3, buf);

              $$ = strdup(buf);
          }
          else if (strcmp($3, "0") == 0)
          {
              printf("// Algebraic Simplification: x - 0 -> x\n");

              $$ = strdup($1);
          }
          else
          {
              sprintf(buf, "%s - %s", $1, $3);
              $$ = strdup(buf);
          }

          free($1);
          free($3);
      }

    | expr '*' expr
      {
          char buf[100];

          if (is_number($1) && is_number($3))
          {
              int result = atoi($1) * atoi($3);

              sprintf(buf, "%d", result);

              printf("// Constant Folding: %s * %s -> %s\n",
                     $1, $3, buf);

              $$ = strdup(buf);
          }
          else if (strcmp($3, "1") == 0)
          {
              printf("// Algebraic Simplification: x * 1 -> x\n");

              $$ = strdup($1);
          }
          else if (strcmp($1, "1") == 0)
          {
              printf("// Algebraic Simplification: 1 * x -> x\n");

              $$ = strdup($3);
          }
          else if (strcmp($3, "2") == 0)
          {
              sprintf(buf, "%s + %s", $1, $1);

              printf("// Strength Reduction: x * 2 -> x + x\n");

              $$ = strdup(buf);
          }
          else
          {
              sprintf(buf, "%s * %s", $1, $3);
              $$ = strdup(buf);
          }

          free($1);
          free($3);
      }

    | expr '/' expr
      {
          char buf[100];

          if (is_number($1) && is_number($3))
          {
              int divisor = atoi($3);

              if (divisor == 0)
              {
                  printf("// Error: Division by zero\n");
                  $$ = strdup("0");
              }
              else
              {
                  int result = atoi($1) / divisor;

                  sprintf(buf, "%d", result);

                  printf("// Constant Folding: %s / %s -> %s\n",
                         $1, $3, buf);

                  $$ = strdup(buf);
              }
          }
          else if (strcmp($3, "1") == 0)
          {
              printf("// Algebraic Simplification: x / 1 -> x\n");

              $$ = strdup($1);
          }
          else
          {
              sprintf(buf, "%s / %s", $1, $3);
              $$ = strdup(buf);
          }

          free($1);
          free($3);
      }
    ;

%%

int main()
{
    printf("Enter Three Address Code statements (end with Ctrl+D):\n");

    yyparse();

    return 0;
}

int yyerror(const char *s)
{
    printf("Syntax Error: %s\n", s);

    return 0;
}
```
